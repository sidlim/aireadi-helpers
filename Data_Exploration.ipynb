```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import exposure, img_as_float
from skimage.filters import frangi, sato
from scipy.ndimage import gaussian_filter, median_filter
import cv2  # optional, for CLAHE and morphology
from skimage.morphology import remove_small_objects, disk, opening

# `enface` should be your 2D numpy array (float or integer)
img = 1 - img_as_float(arr)  # convert to float in [0,1]

# 1) percentile clip & rescale
p_low, p_high = 1, 99
lo, hi = np.percentile(img, (p_low, p_high))
img_clip = np.clip(img, lo, hi)
img_rescale = (img_clip - lo) / (hi - lo + 1e-12)

# 2) small denoise (optional)
img_smooth = median_filter(img_rescale, size=3)  # try gaussian_filter(..., sigma=0.5)

# 3) local contrast (CLAHE) - OpenCV
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
img_8bit = (img_smooth * 255).astype(np.uint8)
img_clahe = clahe.apply(img_8bit)
img_clahe_f = img_clahe.astype(float) / 255.0

# 4) remove slow background (top-hat) - morphological opening
selem = disk(15)   # adjust size for large background trends
bg = opening(img_clahe_f, selem)
img_tophat = img_clahe_f - bg
img_tophat = np.clip(img_tophat, 0, 1)

# 5) vesselness filter (Frangi or Sato)
vessel_frangi = frangi(img_tophat, scale_range=(1,4), scale_step=1, alpha=0.5, beta=0.5)
vessel_sato   = sato(img_tophat, sigmas=range(1,4))

# 6) normalize for display
def norm01(x): return (x - x.min()) / (x.max() - x.min() + 1e-12)
vf = norm01(vessel_frangi)
vs = norm01(vessel_sato)

# 7) threshold + clean if you want binary mask
mask = vf > 0.2           # try different cutoffs or use Otsu
mask = remove_small_objects(mask, min_size=30)

# Display options
plt.figure(figsize=(12,6))
plt.subplot(1,4,1); plt.title("rescaled"); plt.imshow(img_rescale, cmap='gray'); plt.axis('off')
plt.subplot(1,4,2); plt.title("CLAHE+tophat"); plt.imshow(img_tophat, cmap='gray'); plt.axis('off')
plt.subplot(1,4,3); plt.title("Frangi"); plt.imshow(vf, cmap='inferno'); plt.axis('off')
plt.subplot(1,4,4); plt.title("Frangi mask overlay"); 
plt.imshow(img_rescale, cmap='gray'); plt.imshow(mask, cmap='spring', alpha=0.6); plt.axis('off')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import img_as_float
from skimage.filters import frangi, sato, threshold_otsu
from skimage.morphology import white_tophat, black_tophat, disk, remove_small_objects
from scipy.ndimage import gaussian_filter, convolve, rotate

# --- Helpers ---
def norm01(x): return (x - x.min()) / (x.max() - x.min() + 1e-12)

def matched_filter_max(img, length=21, width_sigma=1.0, n_angles=24):
    ks = length
    kernel = np.zeros((ks, ks), dtype=float)
    kernel[ks//2, :] = 1.0
    kernel = gaussian_filter(kernel, sigma=(width_sigma, width_sigma*4))
    kernel /= kernel.sum()
    responses = []
    for angle in np.linspace(0, 180, n_angles, endpoint=False):
        krot = rotate(kernel, angle, reshape=False, order=1)
        responses.append(convolve(img, krot, mode='reflect'))
    return np.max(np.stack(responses), axis=0)

def binarize(im, min_size=50):
    im = norm01(im)
    th = threshold_otsu(im)
    mask = im > th
    mask = remove_small_objects(mask, min_size=min_size)
    return mask

# --- Load image (replace with your enface array) ---
I = img_as_float(arr)

# --- Precompute variants ---
I_inv = 1.0 - I
dog = gaussian_filter(I, 1.0) - gaussian_filter(I, 3.0)
tophat_w = white_tophat(I, disk(15))
tophat_b = black_tophat(I, disk(15))

f_orig = frangi(I, scale_range=(1,8), scale_step=1, alpha=0.5, beta=0.5)
f_inv  = frangi(I_inv, scale_range=(1,8), scale_step=1, alpha=0.5, beta=0.5)
s_orig = sato(I, sigmas=range(1,8))
mf     = matched_filter_max(I, length=21, width_sigma=0.8, n_angles=32)

# --- Images and masks ---
enhancements = [
    ("Original", I),
    ("DoG", dog),
    ("Tophat white", tophat_w),
    ("Tophat black", tophat_b),
    ("Frangi(orig)", f_orig),
    ("Frangi(inv)", f_inv),
    ("Sato(orig)", s_orig),
    ("Matched filter", mf),
]

masks = [(title+" mask", binarize(im)) for title, im in enhancements]

# --- Plot ---
fig, axs = plt.subplots(3, len(enhancements), figsize=(20,10))
for i, (title, im) in enumerate(enhancements):
    axs[0,i].imshow(norm01(im), cmap='inferno')
    axs[0,i].set_title(title, fontsize=10)
    axs[0,i].axis('off')

for i, (title, im) in enumerate(enhancements):
    axs[1,i].imshow(norm01(im), cmap='gray')
    axs[1,i].set_title(title+" (gray)", fontsize=9)
    axs[1,i].axis('off')

for i, (title, mask) in enumerate(masks):
    axs[2,i].imshow(mask, cmap='gray')
    axs[2,i].set_title(title, fontsize=9)
    axs[2,i].axis('off')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io, exposure, filters, morphology, util
from skimage.restoration import denoise_bilateral

# --- 1. Denoise ---
# Option A: Gaussian smoothing
denoised = filters.gaussian(arr, sigma=1)

# Option B: Bilateral denoising (preserves edges, slower)
# denoised = denoise_bilateral(arr, sigma_color=0.05, sigma_spatial=3)

# --- 2. Contrast Enhancement (CLAHE) ---
clahe = exposure.equalize_adapthist(denoised, clip_limit=0.03)

# --- 3. Thresholding ---
# Adaptive thresholding
block_size = 35  # adjust depending on vessel scale
thresh = filters.threshold_local(clahe, block_size=block_size, offset=-0.01)
binary = clahe > thresh

# Clean small specks
binary = morphology.remove_small_objects(binary, min_size=64)
binary = morphology.remove_small_holes(binary, area_threshold=64)

# --- 4. Skeletonization ---
skeleton = morphology.skeletonize(binary)
# Alternatively: skeleton_3d = morphology.skeletonize_3d(binary)

# --- Plotting ---
fig, axes = plt.subplots(1, 4, figsize=(18, 5))
axes[0].imshow(arr, cmap="gray")
axes[0].set_title("Original")
axes[1].imshow(denoised, cmap="gray")
axes[1].set_title("Denoised")
axes[2].imshow(binary, cmap="gray")
axes[2].set_title("Binary mask")
axes[3].imshow(skeleton, cmap="gray")
axes[3].set_title("Skeletonized")
for ax in axes: ax.axis("off")
plt.tight_layout()
plt.show()
```
